```python
import pandas as pd

data = {

    "Name" : ["Ali", "Ahmed", "Ayesha", "Hassan", "Hina"],
    "Roll No" : [1000, 2000, 3000, 4000, 5000],
    "Math" : [80, 90, 70, 60, 50],
    "Science" : [70, 80, 90, 60, 50],
    "English" : [60, 70, 80, 90, 100]

}

results = pd.DataFrame(data)
results
```

```python
results.describe()
```

```python
results = pd.DataFrame(data)
print(results)
print(results.describe())
print("Top Math Scorer:", results.loc[results['Math'].idxmax(), 'Name'])
```
